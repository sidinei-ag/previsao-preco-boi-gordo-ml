# IMPORT

```python
import pandas as pd
import seaborn as sns
```

# LOAD DATA

```python
df_boi = pd.read_csv('../data/raw/cepea_boi_gordo.csv')
df_dolar = pd.read_csv('../data/raw/cepea_dolar.csv', sep=';')
df_milho = pd.read_csv('../data/raw/cepea_milho.csv', sep=';')
df_suino = pd.read_csv('../data/raw/cepea_suinos.csv', sep=';')
```

# EDA & PRE-PROCESSING

### Processamento Inicial

```python
print(f'Quantidade dados dataset df_boi:{len(df_boi)}')
print(f'Quantidade dados dataset df_dolar:{len(df_dolar)}')
print(f'Quantidade dados dataset df_dolar:{len(df_milho)}')
print(f'Quantidade dados dataset df_dolar:{len(df_suino)}')
```

```python
df_milho.head()
```

### Renomeando Colunas

```python
df_boi = df_boi.rename(columns={
    'Data':'data',
    'Boi São Paulo R$/@':'preco_boi_gordo'
})
```

```python
df_dolar = df_dolar.rename(columns={
    'Data':'data',
    'Valor':'cotacao_dolar'
})
```

```python
df_milho = df_milho.rename(columns={
    'Data':'data',
    'Valor Real':'milho_real',
    'Valor Dolar': 'milho_dolar'
})
```

```python
df_suino = df_suino.rename(columns={
    'Data':'data',
    'Valor':'preco_suino'
})
```

### Unindo datasets

```python
# Alterando os formatos de datas

df_suino['data'] = pd.to_datetime(df_suino['data'], format='%d/%m/%Y').dt.date
df_milho['data'] = pd.to_datetime(df_milho['data'], format='%d/%m/%Y').dt.date
df_dolar['data'] = pd.to_datetime(df_dolar['data'], format='%d/%m/%Y').dt.date
df_boi['data'] = pd.to_datetime(df_boi['data'], format='%d/%m/%Y').dt.date
```

```python
df = df_boi.copy()
```

```python
df = df.merge(
    df_suino,
    on='data',
    how='left',
    validate='1:1'
)
```

```python
df = df.merge(
    df_milho,
    on='data',
    how='left',
    validate='1:1'
)
```

```python
df = df.merge(
    df_dolar,
    on='data',
    how='left',
    validate='1:1'
)
```
